```python

# Market Sales Calculator

# Get details of three items
item1 = input("First item name: ")
price1 = float(input("Price: "))
quantity1 = int(input("Quantity: "))

item2 = input("Second item name: ")
price2 = float(input("Price: "))
quantity2 = int(input("Quantity: "))

item3 = input("Third item name: ")
price3 = float(input("Price: "))
quantity3 = int(input("Quantity: "))

# Calculate line totals
total1 = price1 * quantity1
total2 = price2 * quantity2
total3 = price3 * quantity3

# Calculate subtotal
subtotal = total1 + total2 + total3

# Apply discount
if subtotal >= 50000:
    discount = subtotal * 0.10
elif subtotal >= 20000:
    discount = subtotal * 0.05
else:
    discount = 0

# Calculate final total
final_total = subtotal - discount

# Print receipt
print("---------- MARKET RECEIPT ----------")
print(f"{item1}: {quantity1} x NGN{price1:,.2f} = NGN{total1:,.2f}")
print(f"{item2}: {quantity2} x NGN{price2:,.2f} = NGN{total2:,.2f}")
print(f"{item3}: {quantity3} x NGN{price3:,.2f} = NGN{total3:,.2f}")

print("------------------------------------")
print(f"Subtotal: NGN{subtotal:,.2f}")
print(f"Discount: NGN{discount:,.2f}")
print(f"Final Total: NGN{final_total:,.2f}")
print("------------------------------------")

```
